# Chase The Dot - Algorithm Comparison

This notebook visualizes and compares the training metrics across different algorithms (`pid` vs `vpg`).
All metrics are averaged over batches to provide smooth, interpretable learning curves.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Set plot style
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
# Load logs for algorithms
algos = ['pid', 'vpg']
data = {}

for algo in algos:
    file_path = os.path.join("logs", f"{algo}_log.csv")
    if os.path.exists(file_path):
        data[algo] = pd.read_csv(file_path)
        print(f"Loaded {algo} data: {len(data[algo])} records")
    else:
        print(f"{algo} data not found.")
```

```python
if data:
    plt.figure(figsize=(10, 5))
    for algo, df in data.items():
        plt.plot(df['step'], df['avg_distance'], label=algo.upper(), alpha=0.8)
    plt.title('Average Distance (Tracking Error)')
    plt.xlabel('Environment Steps')
    plt.ylabel('Distance (pixels)')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
if data:
    plt.figure(figsize=(10, 5))
    for algo, df in data.items():
        plt.plot(df['step'], df['in_bounds_percent'], label=algo.upper(), alpha=0.8)
    plt.title('Time in Bounds (%)')
    plt.xlabel('Environment Steps')
    plt.ylabel('Percent (%)')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
if data:
    plt.figure(figsize=(10, 5))
    for algo, df in data.items():
        plt.plot(df['step'], df['avg_reward'], label=algo.upper(), alpha=0.8)
    plt.title('Average Reward over Time')
    plt.xlabel('Environment Steps')
    plt.ylabel('Reward')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
if data:
    plt.figure(figsize=(10, 5))
    for algo, df in data.items():
        plt.plot(df['step'], df['loss'], label=algo.upper(), alpha=0.8)
    plt.title('Training Loss')
    plt.xlabel('Environment Steps')
    plt.ylabel('Loss')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
if data:
    plt.figure(figsize=(10, 5))
    for algo, df in data.items():
        plt.plot(df['step'], df['avg_dt'], label=algo.upper(), alpha=0.8)
    plt.title('Network Latency / Frametime')
    plt.xlabel('Environment Steps')
    plt.ylabel('Average dt (seconds)')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

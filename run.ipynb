# Benchmark All Algorithms

This notebook sequentially trains all RL algorithms on the Chase the Dot environment for 250,000 timesteps each.

> **Note:** Make sure you have at least one instance of the `Cy_RL_PS.exe` environment running on port 6102 before starting this notebook!

```python
import os
import gc
from chase_the_dot.run import main

# Configuration
TIMESTEPS = 250000
ALGORITHMS = ["sac", "td3", "ddpg", "ppo", "a2c", "vpg", "pid"]
PORT = 6102

os.makedirs("models", exist_ok=True)
os.makedirs("runs", exist_ok=True)  # TensorBoard logs
```

```python
for algo in ALGORITHMS:
    print(f"\n{'='*60}")
    print(f"Starting {algo.upper()} for {TIMESTEPS} timesteps...")
    print(f"{'='*60}\n")
    
    try:
        # Run natively within the notebook process
        main([
            "--algo", algo,
            "--timesteps", str(TIMESTEPS),
            "--port", str(PORT),
            "--randomize-config"
        ])
    except KeyboardInterrupt:
        print("\nBenchmark interrupted by user.")
        break
    except Exception as e:
        print(f"\n[ERROR] Algorithm {algo.upper()} crashed: {e}")
        break
    
    # Force Python to clear the neural networks from memory before starting the next algorithm
    gc.collect()

print("\nBenchmark complete! Run `uv run tensorboard --logdir runs/` in your terminal to view results.")
```
